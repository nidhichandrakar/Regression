```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Social_Network_Ads.csv')
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 0)
```

## Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
print(X_train)
```

## Training the Logistic Regression model on the Training set

```python
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression(random_state = 0)
classifier.fit(X_train, y_train)
```

## Predicting a new result

```python
print(classifier.predict(sc.transform([[40,87000]])))
```

## Predicting the Test set results

```python
y_pred = classifier.predict(X_test)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

## Making the Confusion Matrix

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix for Test Data:")
print(cm)
accuracy_score(y_test, y_pred)
```

```python
from sklearn.metrics import precision_score, recall_score

precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)

print(f"Precision Score for Test Data : {precision:.2f}")
print(f"Recall Score for Test Data: {recall:.2f}")
```

### Assessing Bias and Variance through Training and Test Accuracy

We can get insights into the bias and variance of our model by comparing its performance on the training set versus the test set.

*   **High Bias (Underfitting):** If the model performs poorly on both the training set and the test set, it likely has high bias. This means the model is too simple and isn't capturing the underlying patterns in the data.
*   **High Variance (Overfitting):** If the model performs very well on the training set but significantly worse on the test set, it likely has high variance. This means the model has learned the training data too specifically, including noise, and doesn't generalize well to new, unseen data.

```python
# Predict on the training set
y_train_pred = classifier.predict(X_train)

# Calculate accuracy on the training set
training_accuracy = accuracy_score(y_train, y_train_pred)*100
print(f"Training Set Accuracy: {training_accuracy:.2f}")
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm_train = confusion_matrix(y_train, y_train_pred)
print("Confusion Matrix for Training Data:")
print(cm_train)
accuracy_train = accuracy_score(y_train, y_train_pred)
print(f"\nAccuracy Score for Training Data: {accuracy_train:.2f}")
```

## Visualising the Testing set ROC curve and AUC

```python
y_pred_proba = classifier.predict_proba(X_test)[:, 1]
print("Predicted probabilities for class 1 on test data:")      # for positive class
display(y_pred_proba[:10])
```

```python
from sklearn.metrics import roc_curve, auc

# Calculate ROC curve && AUC
fpr, tpr, thresholds = roc_curve(y_test, y_pred_proba)
roc_auc = auc(fpr, tpr)
print(f"AUC for Test Data= {roc_auc:.2f}")

# Plot theROC curve
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', label= 'Random Classifier')
#plt.xlim([0.0, 1.0])
#plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()
```

## Visualising the Training set ROC curve and AUC

```python
y_train_pred_proba = classifier.predict_proba(X_train)[:, 1]
print("Predicted probabilities for class 1 on training data:")
display(y_train_pred_proba[:10])
```

```python
# Calculate ROC curve & AUC for training data
fpr_train, tpr_train, thresholds_train = roc_curve(y_train, y_train_pred_proba)
roc_auc_train = auc(fpr_train, tpr_train)
print(f"AUC for Training Data = {roc_auc_train:.2f}")

# Plot the ROC curve for training data
plt.figure(figsize=(8, 6))
plt.plot(fpr_train, tpr_train, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc_train:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', label='Random Classifier')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve (Training Set)')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()
```

Low Bias: Training accuracy of 82.33% indicates that the model is learning the training data reasonably well, suggesting it's not suffering from high bias (underfitting) to a significant degree. It's capturing most of the relevant patterns.

Good Generalization (Low Variance): The fact that test accuracy is even higher than training accuracy is a strong indicator of excellent generalization. Model is not overfitting to the training data; in fact, it appears to perform slightly better on the unseen test data. This means model has low variance and should be quite robust when predicting new, unseen samples.
In summary, these results are generally positive. We've likely achieved a good balance in the bias-variance trade-off, and model appears to generalize very well to new data.

Comparing these two accuracies, you can infer the following:

*   If `Training Accuracy` is significantly higher than `Test Accuracy`, your model might be **overfitting** (high variance).
*   If both `Training Accuracy` and `Test Accuracy` are low, your model might be **underfitting** (high bias).
*   If both accuracies are high and close to each other, your model has a good **bias-variance trade-off**.

## Visualising the Training set results

```python
from matplotlib.colors import ListedColormap
X_set, y_set = sc.inverse_transform(X_train), y_train
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 10, stop = X_set[:, 0].max() + 10, step = 0.5),
                     np.arange(start = X_set[:, 1].min() - 1000, stop = X_set[:, 1].max() + 1000, step = 0.5))
plt.contourf(X1, X2, classifier.predict(sc.transform(np.array([X1.ravel(), X2.ravel()]).T)).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1], c = ListedColormap(('red', 'green'))(i), label = j)
plt.title('Logistic Regression (Training set)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend()
plt.show()
```

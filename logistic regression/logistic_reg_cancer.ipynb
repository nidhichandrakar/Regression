```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
import statsmodels.api as sm

# Convert boolean columns to integer (0 or 1) for VIF calculation
for col in X.select_dtypes(include='bool').columns:
    X[col] = X[col].astype(int)

# Add a constant to the DataFrame for VIF calculation
X_const = sm.add_constant(X)

# Calculate VIF for each feature
vif_data = pd.DataFrame()
vif_data['feature'] = X_const.columns
vif_data['VIF'] = [variance_inflation_factor(X_const.values, i) for i in range(X_const.shape[1])]

# Display the VIF data, sorted by VIF value
print("VIF for Independent Variables:")
display(vif_data.sort_values(by='VIF', ascending=False))

```

```python
# Identify features with low VIF values (e.g., VIF < 10)
low_vif_features = vif_data.loc[vif_data['VIF']<10,].sort_values(by='VIF', ascending=False).reset_index(drop=True)
# Display the features with low VIF values
print("\nFeatures with VIF < 10:\n")
display(low_vif_features)
```

## Splitting the dataset into the Training set and Test set

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 0)
```

## Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
print(X_train)
```

```python
print(X_test)
```

## Training the Logistic Regression model on the Training set

```python
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression(random_state = 0)
classifier.fit(X_train, y_train)
```

## Predicting the Test set results

```python
y_pred = classifier.predict(X_test)
#print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

## Making the Confusion Matrix

## Confusion Matrix and Metrics for Test Data

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix for Test Data:")
print(cm)
accuracy_test = accuracy_score(y_test, y_pred)
print(f"Accuracy Score for Test Data: {accuracy_test:.2f}")
```

```python
from sklearn.metrics import precision_score, recall_score

precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)

print(f"Precision Score for Test Data : {precision:.2f}")
print(f"Recall Score for Test Data: {recall:.2f}")
```

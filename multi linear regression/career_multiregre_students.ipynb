```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant
```

```python
X = data_encoded.drop('Performance Index', axis=1)
```

```python
for col in X.select_dtypes(include=['bool']).columns:
    X[col] = X[col].astype(int)
```

```python
X_const = add_constant(X, has_constant='add') # Use has_constant='add' to prevent adding constant if it already exists

# Prepare an empty DataFrame to store VIF results.
vif_data = pd.DataFrame()
vif_data['feature'] = X_const.columns # Store feature names
# Calculate VIF for each independent variable using a list comprehension.
vif_data['VIF'] = [variance_inflation_factor(X_const.values, i) for i in range(X_const.shape[1])]

print("VIF for Independent Variables:")
display(vif_data.round(2)) # Display VIF values rounded to two decimal places.
```

```python
X = data_encoded.loc[:,['Previous Scores']]
Y = data_encoded.loc[:,'Performance Index']
```

```python
X.head()
```

```python
Y
```

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X,Y)
```

```python
y_pred = regressor.predict(X)
y_pred
```

```python
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error,mean_absolute_percentage_error
mae = mean_absolute_error(Y, y_pred)
print(f"Mean Absolute Error: {mae}")
mse = mean_squared_error(Y, y_pred)
print(f"Mean Squared Error: {mse}")
r_2 = r2_score(Y, y_pred)
print(f"R-squared: {r_2}")
mape = mean_absolute_percentage_error(Y, y_pred)*100
print(f"Mean Absolute Percentage Error: {mape}")
```

```python
import statsmodels.api as sm

# Re-defining X and Y with all features (excluding 'Performance Index')
X_all_features = data_encoded.drop('Performance Index', axis=1)

# Convert boolean columns to int again, as X might have been redefined for single feature earlier
for col in X_all_features.select_dtypes(include=['bool']).columns:
    X_all_features[col] = X_all_features[col].astype(int)

Y_all_features = data_encoded['Performance Index']

# Add a constant to the independent variables for statsmodels
X_all_features_const = sm.add_constant(X_all_features)

# Create and fit the OLS model
model = sm.OLS(Y_all_features, X_all_features_const)
results = model.fit()

# Print the summary of the regression results
print(results.summary())
```
